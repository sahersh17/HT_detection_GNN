```python
def get_chunk_embeddings(code_text, max_length=512, stride=50):
    encoding = tokenizer(
        code_text,
        return_tensors="pt",
        truncation=True,
        max_length=max_length,
        stride=stride,
        return_overflowing_tokens=True,
        padding="max_length",   # <-- pads every chunk to the same length
    )
    num_chunks = encoding["input_ids"].shape[0]
    chunk_embeddings = []
    with torch.no_grad():
        for i in range(num_chunks):
            input_ids = encoding["input_ids"][i].unsqueeze(0).to(device)
            attention_mask = encoding["attention_mask"][i].unsqueeze(0).to(device)
            outputs = model(input_ids=input_ids, attention_mask=attention_mask)

            # masked mean pool — ignore padded positions so the last
            # (shorter) chunk isn't diluted by pad-token embeddings
            hidden = outputs.last_hidden_state.squeeze(0)          # (max_length, 768)
            mask = attention_mask.squeeze(0).unsqueeze(-1).float() # (max_length, 1)
            chunk_emb = (hidden * mask).sum(dim=0) / mask.sum()
            chunk_embeddings.append(chunk_emb.cpu())
    return torch.stack(chunk_embeddings)  # (num_chunks, 768)
```

```python
from transformers import AutoModel, AutoTokenizer
import torch

checkpoint = "microsoft/codebert-base"
device = "cuda"
tokenizer = AutoTokenizer.from_pretrained(checkpoint)
model = AutoModel.from_pretrained(checkpoint).to(device)
model.eval()

def get_token_embeddings(code_text):
    inputs = tokenizer(code_text, return_tensors="pt", truncation=True, max_length=512).to(device)
    with torch.no_grad():
        outputs = model(**inputs)
    # (seq_len, 768) instead of a single pooled (768,) vector
    return outputs.last_hidden_state.squeeze(0).cpu()
```

```python
RTL_DIR = '/content/HT_detection_GNN/trusthub/AES_unzipped'
FEATURES_DIR = '/content/HT_detection_GNN/trusthub/features/data/AES'

def find_rtl_files(benchmark_dir, status_folder_names):
    combined_code = ""
    for root, dirs, files in os.walk(benchmark_dir):
        if any(s.lower() in root.lower() for s in status_folder_names):
            for f in files:
                if f.endswith(('.v', '.sv')):
                    with open(os.path.join(root, f), 'r', errors='ignore') as fh:
                        combined_code += fh.read() + "\n"
    return combined_code

skipped = []
processed = []

for benchmark in sorted(os.listdir(RTL_DIR)):

    if benchmark == "AES_T2200":
        continue
    benchmark_path = os.path.join(RTL_DIR, benchmark, benchmark)
    if not os.path.isdir(benchmark_path):
        continue

    out_dir = os.path.join(FEATURES_DIR, benchmark)
    if not os.path.exists(out_dir):
        skipped.append((benchmark, "no matching features/data folder"))
        continue

    for status, folder_names, out_name in [
        ('clean', ['tjfree'], 'clean_rtl_tokens_chunk.pt'),
        ('trojan', ['tjin', 'tnin'], 'trojan_rtl_tokens_chunk.pt'),
    ]:
        code = find_rtl_files(benchmark_path, folder_names)
        if not code.strip():
            continue
        emb = get_chunk_embeddings(code)
        out_path = os.path.join(out_dir, out_name)
        torch.save(emb, out_path)
        processed.append(out_path)

print(f"Saved {len(processed)} RTL embeddings")
print(f"Skipped {len(skipped)} benchmarks:", skipped[:5])
```

```python
RTL_DIR = '/content/HT_detection_GNN/trusthub/AES_unzipped'  # RS232 folders live here too
FEATURES_DIR = '/content/HT_detection_GNN/trusthub/features/data/RS232'

def find_rtl_files_by_tag(benchmark_dir, tags):
    combined_code = ""
    for root, dirs, files in os.walk(benchmark_dir):
        if any(t.lower() in root.lower() for t in tags):
            for f in files:
                if f.endswith(('.v', '.sv')):
                    with open(os.path.join(root, f), 'r', errors='ignore') as fh:
                        combined_code += fh.read() + "\n"
    return combined_code

skipped = []
processed = []

for benchmark in sorted(os.listdir(RTL_DIR)):
    if not benchmark.startswith('RS232'):
        continue  # skip AES benchmarks, already done

    benchmark_path = os.path.join(RTL_DIR, benchmark, benchmark)
    if not os.path.isdir(benchmark_path):
        continue

    out_dir = os.path.join(FEATURES_DIR, benchmark)
    if not os.path.exists(out_dir):
        skipped.append((benchmark, "no matching features/data folder"))
        continue

    existing_files = os.listdir(out_dir)

    if any('tjfree' in f.lower() for f in existing_files):
        variants = [('clean', ['tjfree'], 'clean_rtl_embedding_chunk.pt'),
                    ('trojan', ['tjin'], 'trojan_rtl_tokens_chunk.pt')]
    else:
        variants = [('trojan', ['src'], 'trojan_rtl_tokens_chunk.pt')]

    for status, folder_names, out_name in variants:
        code = find_rtl_files_by_tag(benchmark_path, folder_names)
        if not code.strip():
            continue
        emb = get_chunk_embeddings(code)
        out_path = os.path.join(out_dir, out_name)
        torch.save(emb, out_path)
        processed.append(out_path)

print(f"Saved {len(processed)} RS232 RTL embeddings")
print(f"Skipped {len(skipped)} benchmarks:", skipped[:5])
```

```python
import os
import json
import torch
from pathlib import Path

FEATURES_ROOT = Path('/content/HT_detection_GNN/trusthub/features')
manifest = json.loads((FEATURES_ROOT / "manifest.json").read_text())

def infer_family(graph_id):
    for fam in ["AES", "RS232"]:
        if graph_id.upper().startswith(fam):
            return fam
    return None

def infer_benchmark(graph_id, family):
    parts = graph_id.split("/")
    for p in parts:
        if p.upper().startswith(family) and "-" in p:
            return p
    return parts[1] if len(parts) > 1 else parts[0]

def infer_status(graph_id, label):
    gid = graph_id.lower()
    if 'tjfree' in gid:
        return 'clean'
    if 'tjin' in gid or 'tnin' in gid:
        return 'trojan'
    return 'clean' if label in (0, '0') else 'trojan' if label in (1, '1') else 'unknown'

expected = set()
for entry in manifest:
    gid = entry["graph_id"]
    family = infer_family(gid)
    if family is None:
        continue
    if 'T2200' in gid:  # known excluded benchmark
        continue
    benchmark = infer_benchmark(gid, family)
    status = infer_status(gid, entry.get("label"))
    if status != 'unknown':
        expected.add((family, benchmark, status))

missing, malformed, ok = [], [], []

for family, benchmark, status in sorted(expected):
    path = FEATURES_ROOT / "data" / family / benchmark / f"{status}_rtl_tokens_chunk.pt"
    if not path.exists():
        missing.append(str(path))
        continue
    try:
        t = torch.load(path)
        if t.dim() != 2 or t.shape[1] != 768:
            malformed.append((str(path), tuple(t.shape)))
        elif torch.isnan(t).any():
            malformed.append((str(path), "contains NaNs"))
        else:
            ok.append((family, benchmark, status, t.shape[0]))
    except Exception as e:
        malformed.append((str(path), str(e)))

print(f"Expected: {len(expected)}   OK: {len(ok)}   Missing: {len(missing)}   Malformed: {len(malformed)}\n")

if missing:
    print("MISSING:")
    for m in missing:
        print(" ", m)

if malformed:
    print("\nMALFORMED:")
    for m in malformed:
        print(" ", m)

print("\nChunk counts (sanity check — should scale with file size, none should be 0):")
for family, benchmark, status, n_chunks in sorted(ok, key=lambda r: -r[3])[:10]:
    print(f"  {family}/{benchmark} ({status}): {n_chunks} chunks")
```
