```python
from transformers import AutoModel, AutoTokenizer
import torch

checkpoint = "microsoft/codebert-base"
device = "cuda"
tokenizer = AutoTokenizer.from_pretrained(checkpoint)
model = AutoModel.from_pretrained(checkpoint).to(device)
model.eval()

def get_embedding(code_text):
    inputs = tokenizer(code_text, return_tensors="pt", truncation=True, max_length=512).to(device)
    with torch.no_grad():
        outputs = model(**inputs)
    return outputs.last_hidden_state.mean(dim=1).squeeze().cpu()  # shape (768,)
```

```python
RTL_DIR = '/content/HT_detection_GNN/trusthub/AES_unzipped'
FEATURES_DIR = '/content/HT_detection_GNN/trusthub/features/data/AES'

def find_rtl_files(benchmark_dir, status_folder_names):
    combined_code = ""
    for root, dirs, files in os.walk(benchmark_dir):
        if any(s.lower() in root.lower() for s in status_folder_names):
            for f in files:
                if f.endswith(('.v', '.sv')):
                    with open(os.path.join(root, f), 'r', errors='ignore') as fh:
                        combined_code += fh.read() + "\n"
    return combined_code

skipped = []
processed = []

for benchmark in sorted(os.listdir(RTL_DIR)):
    benchmark_path = os.path.join(RTL_DIR, benchmark, benchmark)
    if not os.path.isdir(benchmark_path):
        continue

    out_dir = os.path.join(FEATURES_DIR, benchmark)
    if not os.path.exists(out_dir):
        skipped.append((benchmark, "no matching features/data folder"))
        continue

    for status, folder_names, out_name in [
        ('clean', ['tjfree'], 'clean_rtl_embedding.pt'),
        ('trojan', ['tjin', 'tnin'], 'trojan_rtl_embedding.pt'),
    ]:
        code = find_rtl_files(benchmark_path, folder_names)
        if not code.strip():
            continue
        emb = get_embedding(code)
        out_path = os.path.join(out_dir, out_name)
        torch.save(emb, out_path)
        processed.append(out_path)

print(f"Saved {len(processed)} RTL embeddings")
print(f"Skipped {len(skipped)} benchmarks:", skipped[:5])
```

```python
RTL_DIR = '/content/HT_detection_GNN/trusthub/AES_unzipped'  # RS232 folders live here too
FEATURES_DIR = '/content/HT_detection_GNN/trusthub/features/data/RS232'

def find_rtl_files_by_tag(benchmark_dir, tags):
    combined_code = ""
    for root, dirs, files in os.walk(benchmark_dir):
        if any(t.lower() in root.lower() for t in tags):
            for f in files:
                if f.endswith(('.v', '.sv')):
                    with open(os.path.join(root, f), 'r', errors='ignore') as fh:
                        combined_code += fh.read() + "\n"
    return combined_code

skipped = []
processed = []

for benchmark in sorted(os.listdir(RTL_DIR)):
    if not benchmark.startswith('RS232'):
        continue  # skip AES benchmarks, already done

    benchmark_path = os.path.join(RTL_DIR, benchmark, benchmark)
    if not os.path.isdir(benchmark_path):
        continue

    out_dir = os.path.join(FEATURES_DIR, benchmark)
    if not os.path.exists(out_dir):
        skipped.append((benchmark, "no matching features/data folder"))
        continue

    existing_files = os.listdir(out_dir)

    if any('tjfree' in f.lower() for f in existing_files):
        variants = [('clean', ['tjfree'], 'clean_rtl_embedding.pt'),
                    ('trojan', ['tjin'], 'trojan_rtl_embedding.pt')]
    else:
        variants = [('trojan', ['src'], 'trojan_rtl_embedding.pt')]

    for status, folder_names, out_name in variants:
        code = find_rtl_files_by_tag(benchmark_path, folder_names)
        if not code.strip():
            continue
        emb = get_embedding(code)
        out_path = os.path.join(out_dir, out_name)
        torch.save(emb, out_path)
        processed.append(out_path)

print(f"Saved {len(processed)} RS232 RTL embeddings")
print(f"Skipped {len(skipped)} benchmarks:", skipped[:5])
```
